```python
import pandas as pd
```

```python
dataset=pd.read_csv("CKD.csv")
```

```python
dataset
```

```python
dataset=pd.get_dummies(dataset,dtype=int,drop_first=True)
dataset
```

```python
dataset["classification_yes"].value_counts()
```

```python
Independent=dataset[['age', 'bp', 'al', 'su', 'bgr', 'bu', 'sc','sod','pot', 'hrmo', 'pcv',
       'wc', 'rc', 'sg_b', 'sg_c', 'sg_d', 'sg_e', 'rbc_normal', 'pc_normal',
       'pcc_present', 'ba_present', 'htn_yes', 'dm_yes', 'cad_yes',
       'appet_yes', 'pe_yes', 'ane_yes']]
Independent
```

```python
Dependent=dataset[['classification_yes']]
Dependent
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(Independent,Dependent,test_size=0.30,random_state=0)
```

```python
from sklearn.naive_bayes import CategoricalNB
from sklearn.model_selection import GridSearchCV
param_grid={'alpha':[1,5,10,100],'force_alpha':[True,False],'fit_prior':[True,False]}
grid=GridSearchCV(CategoricalNB(),param_grid,refit=True,verbose=3,n_jobs=-1,scoring='f1_weighted')
grid.fit(Independent,Dependent)
```

```python
y_pred=grid.predict(Independent) #(.predict->defined function)
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(Dependent,y_pred)
print(cm)
```

```python
from sklearn.metrics import classification_report
clf_report=classification_report(Dependent,y_pred)
print(clf_report)
```

```python
re=grid.cv_results_
grid_predictions = grid.predict(x_test) 
from sklearn.metrics import f1_score
f1_macro=f1_score(Dependent,y_pred,average='weighted')
print("f1_macro value for {}:".format(grid.best_params_),f1_macro)
from sklearn.metrics import roc_auc_score
roc_auc_score(Dependent,grid.predict_proba(Independent)[:,1])
table=pd.DataFrame.from_dict(re)
table
```
